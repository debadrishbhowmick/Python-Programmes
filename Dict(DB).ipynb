```python
my_dict={
    101:"Amit",
    102:"Rahul",
    103:"Priya",
}
for key in my_dict.keys():
    print(key)
for value in my_dict.values():
    print(value)
for key,value in my_dict.items():
    print(key,value)
```

```python
my_dict = {"a":10,"b":20,"c":30,"d":40}
total = 0
for value in my_dict.values():
    total=total+value
print("Sum of values:",total)
```
