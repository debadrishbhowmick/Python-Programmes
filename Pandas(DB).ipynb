```python
# creating a series from list
import pandas as pd

a = [1,2,3]
myvar = pd.Series(a)
print(myvar)
```

```python
# # creating a series from list with indexes
import pandas as pd

a = [1,2,3]
myvar = pd.Series(a, index = ["x","y","z"])
print(myvar)
```

```python
s1 = pd.Series([10,20,30], index=['a','b','c'], name='Example Series')
print(s1)
```

```python
#Creating a series from a dictionary
s2 = pd.Series({'x':1,'y':2,'z':3})
print(s2)
```

```python
import pandas as pd

calories = {"day1": 420,"day2": 380,"day3": 390}
myvar = pd.Series(calories)
print(myvar)
```

```python
# DATAFRAMES
import pandas as pd

data = {
    "calories": [420,380,390],
    "duration": [50,40,45],
}

myvar = pd.DataFrame(data)
print(myvar)
```

```python
#Locate rows
import pandas as pd

data = {
    "calories": [420,380,390],
    "duration": [50,40,45],
}
#load data into a DataFrame object:
df = pd.DataFrame(data)
print(df.loc[[0,1]])
```

```python
import pandas as pd

data = {
    "calories": [420,380,390],
    "duration": [50,40,45],
}
df = pd.DataFrame(data,index = ["day1","day2","day3"])
print(df)
```

```python
# Preprocessing(CLEAN DATA)
import pandas as pd
 #define a dict with sample data which includes some missing vakues
data = {
    'A':[1,2,3,None,5],
    'B':[None,2,3,4,5],
    'C':[1,2,None,None,5]
}

df = pd.DataFrame(data)
print("original Data:\n",df)
print()
#use dropna() to remove rows with any missing values
df_cleared = df.dropna()
print(df_cleared)
```

```python
#Filling missing values
import pandas as pd
 #define a dict with sample data which includes some missing vakues
data = {
    'A':[1,2,3,None,5],
    'B':[None,2,3,4,5],
    'C':[1,2,None,None,5]
}

df = pd.DataFrame(data)
print("original Data:\n",df)

#filling NaN values with 0
df.fillna(0, inplace=True)
print(df)
```
