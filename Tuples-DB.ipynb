```python
numbers = [(10,20,40),(40,50,60),(70,80,90)]
new_list=[]
for tup in numbers:
    new_tup = tup[:-1] + (100,)
    new_list.append(new_tup)
print('original:',numbers)
print('new:',new_list)
```
